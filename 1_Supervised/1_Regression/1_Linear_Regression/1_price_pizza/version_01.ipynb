```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
# داده‌ها: اندازه پیتزا (اینچ) و قیمت (هزار تومان)
X = np.array([[6], [8], [10], [12]])   # ورودی باید دوبعدی باشد
y = np.array([30, 40, 50, 60])          # خروجی
```

```python
# ساخت و آموزش مدل
model = LinearRegression()
model.fit(X, y)
```

```python
# نمایش شیب و عرض از مبدأ
print("شیب (a):", model.coef_[0])
print("عرض از مبدأ (b):", model.intercept_)
```

```python
# پیش‌بینی برای پیتزای ۱۴ اینچی
اندازه_جدید = np.array([[14]])
پیش‌بینی = model.predict(اندازه_جدید)
print("قیمت پیش‌بینی‌شده برای ۱۴ اینچ:", پیش‌بینی[0], "هزار تومان")
```

```python
# ============================================================
#  فایل پیش‌بینی: predict.py
#  وظیفه: بارگذاری مدل ذخیره‌شده و پیش‌بینی برای ورودی جدید
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import joblib
```

```python
# ---------- ۱. بارگذاری مدل ----------
MODEL_PATH = "pizza_model.pkl"
model = joblib.load(MODEL_PATH)

print("=" * 50)
print("مدل با موفقیت بارگذاری شد.")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۲. پیش‌بینی برای ورودی جدید ----------
اندازه_جدید = 14
قیمت_پیش‌بینی = model.predict([[اندازه_جدید]])[0]

print(f"اندازه پیتزا: {اندازه_جدید} اینچ")
print(f"قیمت پیش‌بینی‌شده: {قیمت_پیش‌بینی:.2f} هزار تومان")
print("=" * 50)
```

```python
# ---------- ۳. رسم نمودار همراه با نقطه پیش‌بینی ----------
# داده‌های اصلی فقط برای رسم نمودار
X = np.array([[6], [8], [10], [12]])
y = np.array([32, 38, 52, 58])

plt.figure(figsize=(8, 5))
```

```python
# نقاط داده واقعی
plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')
```

```python
# خط رگرسیون
X_line = np.linspace(5, 15, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')
```

```python
# نقطه پیش‌بینی‌شده
plt.scatter([اندازه_جدید], [قیمت_پیش‌بینی], color='green',
            marker='*', s=250,
            label=f'پیش‌بینی {اندازه_جدید} اینچ ({قیمت_پیش‌بینی:.1f})')

plt.xlabel('اندازه پیتزا (اینچ)')
plt.ylabel('قیمت (هزار تومان)')
plt.title('پیش‌بینی با مدل ذخیره‌شده')
plt.legend()
plt.grid(True)
```

```python
# ذخیره نمودار
PLOT_PATH = "prediction_plot.png"
plt.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار پیش‌بینی در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

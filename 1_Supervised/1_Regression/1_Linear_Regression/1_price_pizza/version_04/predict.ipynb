```python
# ============================================================
#  فایل پیش‌بینی: predict.ipynb
#  وظیفه: بارگذاری مدل ذخیره‌شده و پیش‌بینی برای ورودی جدید
# ============================================================
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
import joblib
import matplotlib.pyplot as plt
```

```python
# ---------- ۱. بارگذاری مدل ----------
MODEL_PATH = "model/pizza_model.pkl"
model = joblib.load(MODEL_PATH)

print("=" * 50)
print("مدل با موفقیت بارگذاری شد.")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۲. پیش‌بینی برای ورودی جدید ----------
new_size = 14
forecasted_price = model.predict([[new_size]])[0]

print(f"اندازه پیتزا: {new_size} اینچ")
print(f"قیمت پیش‌بینی‌شده: {forecasted_price:.2f} هزار تومان")
#print(f"قیمت پیش‌بینی‌شده: {قیمت_پیش‌بینی:.2f} هزار تومان")
print("=" * 50)
```

```python
# ---------- ۳. رسم نمودار همراه با نقطه پیش‌بینی ----------
# داده‌های اصلی فقط برای رسم نمودار
X = np.array([[6], [8], [10], [12]])
y = np.array([32, 38, 52, 58])

plt.figure(figsize=(8, 5))
# نقاط داده واقعی
plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')
```

```python
# خط رگرسیون
X_line = np.linspace(5, 15, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')
```

```python
# نقطه پیش‌بینی‌شده
plt.scatter([new_size], [forecasted_price], color='green',
            marker='*', s=250,
            label=f'اینچ {new_size} قیمت پیش بینی ({forecasted_price:.2f})')
            
plt.xlabel('اندازه پیتزا (اینچ)')
plt.ylabel('قیمت (هزار تومان)')
plt.title('پیش‌بینی با مدل ذخیره‌شده')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
# ذخیره نمودار
PLOT_PATH = "result/prediction_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار پیش‌بینی در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

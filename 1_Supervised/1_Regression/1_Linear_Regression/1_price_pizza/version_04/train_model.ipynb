```python
# ============================================================
#  فایل آموزش مدل: train_model.ipynb
#  وظیفه: آموزش رگرسیون خطی، ذخیره مدل و رسم نمودار
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import joblib
```

```python
# ---------- ۱. تعریف داده‌ها ----------
# اندازه پیتزا (اینچ)
X = np.array([[6], [8], [10], [12]])

# قیمت (هزار تومان)
y = np.array([32, 38, 52, 58])
```

```python
# ---------- ۲. ساخت و آموزش مدل ----------
model = LinearRegression()
model.fit(X, y)

print("=" * 50)
print("نتایج آموزش مدل")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۳. ذخیره مدل آموزش‌دیده ----------
MODEL_PATH = "model/pizza_model.pkl"
joblib.dump(model, MODEL_PATH)
print(f"مدل با موفقیت در فایل «{MODEL_PATH}» ذخیره شد.")
```

```python
# ---------- ۴. رسم نمودار خط رگرسیون ----------
plt.figure(figsize=(8, 5))
# نقاط داده واقعی
plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')
```

```python
# خط رگرسیون
X_line = np.linspace(5, 15, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

plt.xlabel('اندازه پیتزا (اینچ)')
plt.ylabel('قیمت (هزار تومان)')
plt.title('رگرسیون خطی: قیمت پیتزا بر اساس اندازه')
plt.legend()
plt.grid(True)
fig = plt.gcf()          # دریافت figure فعلی
```

```python
# ذخیره نمودار
PLOT_PATH = "result/regression_plot.png"

plt.show()
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار در فایل «{PLOT_PATH}» ذخیره شد.")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
# داده‌های واقعی‌تر
X = np.array([[6], [8], [10], [12]])
y = np.array([32, 38, 52, 58])
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
print("شیب (a):", round(model.coef_[0], 2))
print("عرض از مبدأ (b):", round(model.intercept_, 2))
```

```python
# پیش‌بینی برای ۱۴ اینچ
print("قیمت پیش‌بینی‌شده برای ۱۴ اینچ:",
      round(model.predict([[14]])[0], 2), "هزار تومان")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([[6], [8], [10], [12]])
y = np.array([32, 38, 52, 58])
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
# رسم نقاط داده
plt.scatter(X, y, color='blue', label='داده‌های واقعی')
```

```python
# رسم خط رگرسیون
X_line = np.linspace(5, 15, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', label='خط رگرسیون')
```

```python
# پیش‌بینی نقطه ۱۴ اینچ
plt.scatter([14], model.predict([[14]]), color='green', marker='*', s=150,
            label='پیش‌بینی ۱۴ اینچ')

plt.xlabel('اندازه پیتزا (اینچ)')
plt.ylabel('قیمت (هزار تومان)')
plt.title('رگرسیون خطی: قیمت پیتزا بر اساس اندازه')
plt.legend()
plt.grid(True)
plt.show()
```

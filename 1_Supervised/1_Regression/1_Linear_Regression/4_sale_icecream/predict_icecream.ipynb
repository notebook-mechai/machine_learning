```python
# ============================================================
#  فایل پیش‌بینی: predict_icecream.py
#  وظیفه: بارگذاری مدل ذخیره‌شده و پیش‌بینی فروش برای دمای جدید
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import joblib
```

```python
# ---------- ۱. بارگذاری مدل ----------
MODEL_PATH = "model/icecream_model.pkl"
model = joblib.load(MODEL_PATH)

print("=" * 50)
print("مدل با موفقیت بارگذاری شد.")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۲. پیش‌بینی برای ورودی جدید ----------
new_temperature = 41
sales_forecasting = model.predict([[new_temperature]])[0]

print(f"دما: {new_temperature} درجه سانتی‌گراد")
print(f"فروش پیش‌بینی‌شده: {sales_forecasting:.2f} عدد")
print("=" * 50)
```

```python
# ---------- ۳. رسم نمودار همراه با نقطه پیش‌بینی ----------
# داده‌های اصلی فقط برای رسم نمودار
X = np.array([[20], [25], [30], [35]])
y = np.array([100, 150, 200, 250])

plt.figure(figsize=(8, 5))

plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')

X_line = np.linspace(10, 45, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

plt.scatter([new_temperature], [sales_forecasting], color='green',
            marker='*', s=250,
            label=f'پیش‌بینی {new_temperature}°C ({sales_forecasting:.0f})')

plt.xlabel('دما (درجه سانتی‌گراد)')
plt.ylabel('فروش بستنی (تعداد)')
plt.title('پیش‌بینی فروش بستنی با مدل ذخیره‌شده')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
PLOT_PATH = "results/icecream_prediction_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار پیش‌بینی در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

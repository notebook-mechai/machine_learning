```python
# ============================================================
#  فایل آموزش مدل: train_icecream_model.py
#  وظیفه: آموزش رگرسیون خطی برای پیش‌بینی فروش بستنی بر اساس دما
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import joblib
```

```python
# ---------- ۱. تعریف داده‌ها ----------
# دما (درجه سانتی‌گراد) — ورودی باید دوبعدی باشد
X = np.array([[20], [25], [30], [35]])

# فروش بستنی (تعداد)
y = np.array([100, 150, 200, 250])
```

```python
# ---------- ۲. ساخت و آموزش مدل ----------
model = LinearRegression()
model.fit(X, y)

print("=" * 50)
print("نتایج آموزش مدل")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۳. ذخیره مدل آموزش‌دیده ----------
MODEL_PATH = "model/icecream_model.pkl"
joblib.dump(model, MODEL_PATH)
print(f"مدل با موفقیت در فایل «{MODEL_PATH}» ذخیره شد.")
```

```python
# ---------- ۴. رسم نمودار خط رگرسیون ----------
plt.figure(figsize=(8, 5))

# نقاط داده واقعی
plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')

# خط رگرسیون
X_line = np.linspace(10, 45, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

# نقطه برخورد خط با محور افقی (دمایی که فروش صفر می‌شود)
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.axvline(-model.intercept_ / model.coef_[0], color='orange',
            linestyle=':', linewidth=1.5,
            label=f'دمای فروش صفر ≈ {-model.intercept_ / model.coef_[0]:.0f}°C')

plt.xlabel('دما (درجه سانتی‌گراد)')
plt.ylabel('فروش بستنی (تعداد)')
plt.title('رگرسیون خطی: فروش بستنی بر اساس دما')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
# ذخیره نمودار (قبل از show)
PLOT_PATH = "results/icecream_regression_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

```python
# ============================================================
#  فایل آموزش مدل: train_fuel_model.py
#  وظیفه: آموزش رگرسیون خطی برای پیش‌بینی مصرف بنزین بر اساس مسافت
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import joblib
```

```python
# ---------- ۱. تعریف داده‌ها ----------
# مسافت (کیلومتر) — ورودی باید دوبعدی باشد
X = np.array([[100], [200], [300], [400]])

# مصرف بنزین (لیتر)
y = np.array([8, 16, 24, 32])
```

```python
# ---------- ۲. ساخت و آموزش مدل ----------
model = LinearRegression()
model.fit(X, y)

print("=" * 50)
print("نتایج آموزش مدل")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.4f}")
print(f"عرض از مبدأ (b): {model.intercept_:.4f}")
print("=" * 50)
```

```python
# ---------- ۳. ذخیره مدل آموزش‌دیده ----------
MODEL_PATH = "model/fuel_model.pkl"
joblib.dump(model, MODEL_PATH)
print(f"مدل با موفقیت در فایل «{MODEL_PATH}» ذخیره شد.")
```

```python
# نقاط داده واقعی
plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')

# خط رگرسیون
X_line = np.linspace(0, 450, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

plt.xlabel('مسافت (کیلومتر)')
plt.ylabel('مصرف بنزین (لیتر)')
plt.title('رگرسیون خطی: مصرف بنزین بر اساس مسافت')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
# ذخیره نمودار (قبل از show)
PLOT_PATH = "results/fuel_regression_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

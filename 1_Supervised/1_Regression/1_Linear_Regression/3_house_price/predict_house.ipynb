```python
# ============================================================
#  فایل پیش‌بینی: predict_house.py
#  وظیفه: بارگذاری مدل ذخیره‌شده و پیش‌بینی قیمت برای متراژ جدید
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import joblib
```

```python
# ---------- ۱. بارگذاری مدل ----------
MODEL_PATH = "model/house_model.pkl"
model = joblib.load(MODEL_PATH)

print("=" * 50)
print("مدل با موفقیت بارگذاری شد.")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۲. پیش‌بینی برای ورودی جدید ----------
new_area = 90
forecasted_price = model.predict([[new_area]])[0]

print(f"متراژ: {new_area} متر مربع")
print(f"قیمت پیش‌بینی‌شده: {forecasted_price:0.2f} میلیون تومان")
print("=" * 50)
```

```python
# ---------- ۳. رسم نمودار همراه با نقطه پیش‌بینی ----------
# داده‌های اصلی فقط برای رسم نمودار
X = np.array([[50], [80], [100], [120]])
y = np.array([2500, 4000, 5000, 6000])

plt.figure(figsize=(8, 5))

plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')

X_line = np.linspace(40, 130, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

plt.scatter([new_area], [forecasted_price], color='green',
            marker='*', s=250,
            label = f'متر {new_area} پیش بینی ({forecasted_price:0.2f})')
            
plt.xlabel('متراژ (متر مربع)')
plt.ylabel('قیمت (میلیون تومان)')
plt.title('پیش‌بینی قیمت خانه با مدل ذخیره‌شده')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
PLOT_PATH = "results/house_prediction_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار پیش‌بینی در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

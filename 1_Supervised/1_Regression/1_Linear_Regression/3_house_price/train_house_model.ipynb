```python
# ============================================================
#  فایل آموزش مدل: train_house_model.ipynb
#  وظیفه: آموزش رگرسیون خطی برای پیش‌بینی قیمت خانه، ذخیره مدل و رسم نمودار
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import joblib
```

```python
# ---------- ۱. تعریف داده‌ها ----------
# متراژ خانه (متر مربع) — ورودی باید دوبعدی باشد
X = np.array([[50], [80], [100], [120]])

# قیمت خانه (میلیون تومان)
y = np.array([2500, 4000, 5000, 6000])
```

```python
# ---------- ۲. ساخت و آموزش مدل ----------
model = LinearRegression()
model.fit(X, y)

print("=" * 50)
print("نتایج آموزش مدل")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۳. ذخیره مدل آموزش‌دیده ----------
MODEL_PATH = "model/house_model.pkl"
joblib.dump(model, MODEL_PATH)
print(f"مدل با موفقیت در فایل «{MODEL_PATH}» ذخیره شد.")
```

```python
# خط رگرسیون
X_line = np.linspace(40, 130, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

plt.xlabel('متراژ (متر مربع)')
plt.ylabel('قیمت (میلیون تومان)')
plt.title('رگرسیون خطی: قیمت خانه بر اساس متراژ')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
# ذخیره نمودار (قبل از show)
PLOT_PATH = "results/house_regression_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

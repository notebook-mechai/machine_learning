```python
# ============================================================
#  فایل آموزش مدل: train_salary_model.ipynb
#  وظیفه: آموزش رگرسیون خطی برای پیش‌بینی حقوق، ذخیره مدل و رسم نمودار
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import joblib
```

```python
# ---------- ۱. تعریف داده‌ها ----------
# سابقه کار (سال) — ورودی باید دوبعدی باشد
X = np.array([[1], [2], [3], [4], [5]])

# حقوق (میلیون تومان)
y = np.array([5, 8, 11, 14, 17])

# ---------- ۲. ساخت و آموزش مدل ----------
model = LinearRegression()
model.fit(X, y)

print("=" * 50)
print("نتایج آموزش مدل")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۳. ذخیره مدل آموزش‌دیده ----------
MODEL_PATH = "model/salary_model.pkl"
joblib.dump(model, MODEL_PATH)
print(f"مدل با موفقیت در فایل «{MODEL_PATH}» ذخیره شد.")
```

```python
# ---------- ۴. رسم نمودار خط رگرسیون ----------
plt.figure(figsize=(8, 5))

# نقاط داده واقعی
plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')

# خط رگرسیون
X_line = np.linspace(0, 7, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

plt.xlabel('سابقه کار (سال)')
plt.ylabel('حقوق (میلیون تومان)')
plt.title('رگرسیون خطی: حقوق بر اساس سابقه کار')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
# ذخیره نمودار (قبل از show)
PLOT_PATH = "results/salary_regression_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```

```python
# ============================================================
#  فایل پیش‌بینی: predict_salary.py
#  وظیفه: بارگذاری مدل ذخیره‌شده و پیش‌بینی حقوق برای سابقه جدید
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import joblib
```

```python
# ---------- ۱. بارگذاری مدل ----------
MODEL_PATH = "model/salary_model.pkl"
model = joblib.load(MODEL_PATH)

print("=" * 50)
print("مدل با موفقیت بارگذاری شد.")
print("=" * 50)
print(f"شیب (a): {model.coef_[0]:.2f}")
print(f"عرض از مبدأ (b): {model.intercept_:.2f}")
print("=" * 50)
```

```python
# ---------- ۲. پیش‌بینی برای ورودی جدید ----------
new_record = 6
forecasted_salary = model.predict([[new_record]])[0]

print(f"سابقه کار: {new_record} سال")
print(f"حقوق پیش‌بینی‌شده: {forecasted_salary:.2f} میلیون تومان")
print("=" * 50)
```

```python
# ---------- ۳. رسم نمودار همراه با نقطه پیش‌بینی ----------
# داده‌های اصلی فقط برای رسم نمودار
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([5, 8, 11, 14, 17])

plt.figure(figsize=(8, 5))

plt.scatter(X, y, color='blue', s=80, label='داده‌های واقعی')

X_line = np.linspace(0, 7, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='red', linewidth=2, label='خط رگرسیون')

plt.scatter([new_record], [forecasted_salary], color='green',
            marker='*', s=250,
            label=f'سال {new_record} پیش بینی ({forecasted_salary:.2f})')
            #label=f'پیش‌بینی {سابقه_جدید} سال ({حقوق_پیش‌بینی:.1f})')

plt.xlabel('سابقه کار (سال)')
plt.ylabel('حقوق (میلیون تومان)')
plt.title('پیش‌بینی حقوق با مدل ذخیره‌شده')
plt.legend()
plt.grid(True)
fig = plt.gcf()
```

```python
PLOT_PATH = "results/salary_prediction_plot.png"
fig.savefig(PLOT_PATH, dpi=150, bbox_inches='tight')
print(f"نمودار پیش‌بینی در فایل «{PLOT_PATH}» ذخیره شد.")

plt.show()
```
